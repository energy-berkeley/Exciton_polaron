We have our own result in real space. First we need to cross check with Shastry and Mattis. They calculated the down spin spectral weight, J>0, k = 0. It should have a delta pea at omega/J + s/2 = 3.5 at Hatree, or no S+ and S-.. |J| = 1 unit, eps(k)/W the tight binding/bandwidth = 0, J, 3J.

For the J<0 case, omega/J ~ -2 will have a polaron bound state. Almost delta over there. The energy of such bound state should be below the Hatree value and in the weak coupling limit J<<W, where W is the bandwidth of conduction electron. 

First we will just find the electron propagator.
Aplus = 
Aminus = 
rho = (-(z-J0-J1/4) + np.sqrt(()**2 -4Aplus*Aminus))/2Aminus
G(z,K) = 1/(z+2tcosK+J1/4-J1**2*gamma/4 * 1/(z-J0+J1/3+2(texp(-iK)+J0/2)*rho))

where z = omega + i eta.

Then set K = 0, |J| = 1, t = 0, 0.5, 3/2 (esp(K)/W is varied in the paper because they assmed a parabolica diserso Wk**2, we can cross check by just changing 2tcosK into Wk**2 ) and find imaginary part of thr above propagator, plot the spectral weight aganist omega +1/4

```python
"""
Reproduce Shastry-Mattis (1981) magnetic polaron figures
using our corrected parameter notation:
- J₀: S·S interactions (spin-spin coupling)
- J₁: Sᶻsᶻ diagonal coupling  
- γJ₁: S⁺s⁻ flip coupling with anisotropy γ
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams

# Set publication-quality plot parameters
rcParams['font.size'] = 11
rcParams['font.family'] = 'serif'
rcParams['axes.labelsize'] = 11
rcParams['xtick.labelsize'] = 10
rcParams['ytick.labelsize'] = 10

def green_function_down_spin(omega, K, t, J0, J1, gamma, s=0.5, eta=1e-4):
    """
    Calculate down-spin Green's function G↓(z,K)
    
    Parameters from our corrected notation:
    - J₀: magnon hopping and on-site (S·S terms)
    - J₁: diagonal Sᶻsᶻ coupling
    - γ: anisotropy parameter for flip terms
    - s: localized spin quantum number (s=1/2 for us)
    """
    # Complex frequency with damping
    z = omega + 1j * eta
    
    # Define parameters
    A_plus = t * np.exp(1j * K) + J0 / 2
    A_minus = t * np.exp(-1j * K) + J0 / 2
    
    # α and β from equations of motion
    alpha = z - J0 - J1 / 4
    beta = z - J0 + J1 / 4
    
    # Propagation ratio ρ(z,K)
    discriminant = alpha**2 - 4 * A_plus * A_minus
    rho = (-alpha + np.sqrt(discriminant)) / (2 * A_minus)
    
    # Down-spin Green's function
    # G↓(z,K,-) = 1 / [z + 2t·cos(K) + J₁/4 - (γ²J₁²/4)/(β + 2A₋ρ)]
    epsilon_k = -2 * t * np.cos(K)  # tight-binding dispersion
    
    self_energy = (gamma**2 * J1**2 / 4) / (beta + 2 * A_minus * rho)
    
    G_down = 1 / (z - epsilon_k + J1 / 4 - self_energy)
    
    return G_down

def spectral_weight(omega, K, t, J0, J1, gamma, s=0.5, eta=1e-4):
    """
    Calculate spectral weight: ρ↓(k,ω) = -(1/π)Im[G↓(ω,k)]
    """
    G = green_function_down_spin(omega, K, t, J0, J1, gamma, s, eta)
    return -np.imag(G) / np.pi

def reproduce_fig2_J_positive(J_mag=1.0, s=0.5, gamma=1.0):
    """
    Reproduce Fig 2: Down-spin spectral weight for J>0, k=0
    
    In Shastry-Mattis notation with s=7/2:
    - Hartree peak at ω/J + s/2 ≈ 3.5
    
    For our case with s=1/2:
    - We need to match their physics
    """
    # Set k=0 (band bottom)
    K = 0.0
    
    # For J>0 case (ferromagnetic coupling)
    # Setting J₀ = J₁ = J_mag to match Shastry-Mattis unified J
    J0 = J_mag
    J1 = J_mag
    
    # Electron hopping t (they effectively set t=0 at k=0)
    t = 0.0
    
    # Small damping for sharp features
    eta = 0.01
    
    # Omega range: plot vs ω/J + s/2
    omega_over_J = np.linspace(-2, 5, 1000)
    omega = omega_over_J * J_mag
    
    # Calculate spectral weight
    rho = spectral_weight(omega, K, t, J0, J1, gamma, s, eta)
    
    # Plot
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(omega_over_J + s/2, rho, 'b-', linewidth=2, label='Exact')
    
    # Mark Hartree position
    # Hartree: ω = ε_k + Js/2 = 0 + J*s/2
    hartree_omega = J_mag * s / 2
    hartree_x = hartree_omega / J_mag + s/2
    ax.axvline(hartree_x, color='r', linestyle='--', linewidth=1.5, 
               label=f'Hartree (ω/J+s/2={hartree_x:.2f})')
    
    ax.set_xlabel(r'$\omega/J + s/2$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Down-spin spectral weight: J>0, k=0\n(γ={gamma}, s={s})', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_xlim([omega_over_J[0] + s/2, omega_over_J[-1] + s/2])
    ax.set_ylim(bottom=0)
    
    plt.tight_layout()
    return fig

def reproduce_fig3_J_positive_varied_k(J_mag=1.0, W=10.0, s=0.5, gamma=1.0):
    """
    Reproduce Fig 3: Down-spin spectral weight for J>0 at various ε_k/W
    
    They varied ε_k = 0, J, 3J corresponding to different k values
    """
    K = 0.0  # We'll vary t to change ε_k = -2t cos(K) = -2t
    
    J0 = J_mag
    J1 = J_mag
    eta = 0.03
    
    # Different band energies: ε_k = 0, J, 3J
    epsilon_values = [0, J_mag, 3*J_mag]
    t_values = [0, -J_mag/2, -3*J_mag/2]  # Since ε_k = -2t at K=0
    
    fig, ax = plt.subplots(figsize=(10, 6))
    
    colors = ['blue', 'green', 'red']
    
    for idx, (eps_k, t_val) in enumerate(zip(epsilon_values, t_values)):
        omega_over_J = np.linspace(-2, 10, 1000)
        omega = omega_over_J * J_mag
        
        rho = spectral_weight(omega, K, t_val, J0, J1, gamma, s, eta)
        
        label = f'$\\varepsilon_k = {eps_k/J_mag:.0f}J$'
        ax.plot(omega_over_J, rho, color=colors[idx], linewidth=2, label=label)
        
        # Hartree position for this ε_k
        hartree_omega = eps_k + J_mag * s / 2
        hartree_x = hartree_omega / J_mag
        ax.axvline(hartree_x, color=colors[idx], linestyle='--', 
                   linewidth=1, alpha=0.5)
    
    ax.set_xlabel(r'$\omega/J + s/2$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Down-spin spectral weight: J>0, varying $\\varepsilon_k$\n(γ={gamma}, s={s})', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim(bottom=0)
    
    plt.tight_layout()
    return fig

def reproduce_fig5_J_negative(J_mag=1.0, s=0.5, gamma=1.0):
    """
    Reproduce Fig 5: Down-spin spectral weight for J<0, k=0
    Polaronic bound state case
    
    For J<0, bound state at ω/J ≈ -2 (below Hartree)
    """
    K = 0.0
    
    # For J<0 case (antiferromagnetic coupling)
    J0 = -J_mag
    J1 = -J_mag
    t = 0.0
    
    # Very small damping to show delta-like bound state
    eta = 0.005
    
    omega_over_J = np.linspace(-3, 4, 2000)
    omega = omega_over_J * J_mag
    
    rho = spectral_weight(omega, K, t, J0, J1, gamma, s, eta)
    
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(omega_over_J, rho, 'b-', linewidth=2, label='Exact (bound state)')
    
    # Hartree position (should be above bound state)
    hartree_omega = J_mag * s / 2  # Note: J_mag is positive, J0/J1 are negative
    hartree_x = hartree_omega / J_mag
    ax.axvline(hartree_x, color='r', linestyle='--', linewidth=1.5, 
               label=f'Hartree (ω/J={hartree_x:.2f})')
    
    ax.set_xlabel(r'$\omega/J$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Down-spin spectral weight: J<0 (polaron), k=0\n(γ={gamma}, s={s})', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim(bottom=0)
    
    # Add annotation for bound state
    bound_state_x = -2.0  # Expected around ω/J ≈ -2
    ax.annotate('Bound state\n(polaron)', xy=(bound_state_x, ax.get_ylim()[1]*0.8),
                xytext=(bound_state_x + 1, ax.get_ylim()[1]*0.6),
                arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
                fontsize=10, color='red')
    
    plt.tight_layout()
    return fig

def compare_all_figures():
    """
    Generate all three comparison figures
    """
    # Parameters matching Shastry-Mattis as closely as possible
    # They use J=0.2eV, W=2eV, but we normalize to |J|=1
    
    print("Generating Figure 2 (J>0, k=0)...")
    fig2 = reproduce_fig2_J_positive(J_mag=1.0, s=0.5, gamma=1.0)
    fig2.savefig('fig2_J_positive_k0.pdf', dpi=300, bbox_inches='tight')
    
    print("Generating Figure 3 (J>0, varied ε_k)...")
    fig3 = reproduce_fig3_J_positive_varied_k(J_mag=1.0, W=10.0, s=0.5, gamma=1.0)
    fig3.savefig('fig3_J_positive_varied_k.pdf', dpi=300, bbox_inches='tight')
    
    print("Generating Figure 5 (J<0, polaron)...")
    fig5 = reproduce_fig5_J_negative(J_mag=1.0, s=0.5, gamma=1.0)
    fig5.savefig('fig5_J_negative_polaron.pdf', dpi=300, bbox_inches='tight')
    
    print("\nAll figures saved!")
    
    plt.show()

if __name__ == "__main__":
    compare_all_figures()
```

```python
"""
Reproduce Shastry-Mattis (1981) magnetic polaron figures
Following the paper's notation and using parabolic dispersion: ε_k = W k²

Key corrections:
1. Use parabolic dispersion ε_k = W k² instead of tight-binding
2. Use parabolic magnon dispersion ω_q = w_m q²
3. Match paper's parameters: |J|=1, W=10, w_m=0.01, s=7/2
4. Fix division warnings by adding proper regularization
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
from scipy import integrate

# Set publication-quality plot parameters
rcParams['font.size'] = 11
rcParams['font.family'] = 'serif'
rcParams['axes.labelsize'] = 11
rcParams['xtick.labelsize'] = 10
rcParams['ytick.labelsize'] = 10

def compute_A_function(omega, k, W, J, wm, s=0.5, eta=1e-4, q_max=1.0, nq=500):
    """
    Calculate A(k,ω) from Eq. (13) in paper using parabolic dispersions.
    
    A(k,ω) = (1/N) Σ_q 1/[ε_{k-q} + ω_q - Js/2 - ω]
    
    With parabolic forms:
    ε_k = W k²
    ω_q = w_m q²
    
    Parameters:
    -----------
    omega : float or array
        Frequency
    k : float
        Electron momentum
    W : float
        Electron bandwidth parameter
    J : float
        Coupling constant (signed)
    wm : float
        Maximum magnon energy parameter  
    s : float
        Spin quantum number
    eta : float
        Small imaginary part for regularization
    q_max : float
        Maximum q value for integration (in units where BZ goes to 1)
    nq : int
        Number of q points for integration
    """
    z = omega + 1j * eta
    
    # For 1D with parabolic dispersion, integrate over q
    # We'll use a discrete sum to approximate the integral
    q_vals = np.linspace(-q_max, q_max, nq)
    dq = q_vals[1] - q_vals[0]
    
    # Calculate ε_{k-q} = W(k-q)²
    epsilon_kmq = W * (k - q_vals)**2
    
    # Calculate ω_q = w_m q²
    omega_q = wm * q_vals**2
    
    # Denominator: ε_{k-q} + ω_q - Js/2 - ω - iη
    denominator = epsilon_kmq + omega_q - J*s/2 - z
    
    # Sum (integral approximation): (1/N) Σ_q → ∫dq/(2π) in continuum limit
    # For discrete: multiply by dq/(2π) to approximate integral
    A = np.sum(1.0 / denominator) * dq / (2 * np.pi)
    
    return A

def self_energy(omega, k, W, J, wm, gamma, s=0.5, eta=1e-4):
    """
    Calculate self-energy Σ(k,ω) from Eq. (33):
    
    Σ(k,ω) = -γ² (J²s/2) × A(k,ω) / [1 + (J/2)A(k,ω)]
    
    Parameters:
    -----------
    omega : float or array
        Frequency
    k : float
        Electron momentum  
    W : float
        Bandwidth
    J : float
        Coupling (signed)
    wm : float
        Max magnon energy
    gamma : float
        Anisotropy parameter
    s : float
        Spin quantum number
    eta : float
        Regularization
    """
    A = compute_A_function(omega, k, W, J, wm, s, eta)
    
    # Σ = -γ² (J²s/2) × A / [1 + (J/2)A]
    numerator = -gamma**2 * (J**2 * s / 2) * A
    denominator = 1 + (J / 2) * A
    
    # Add small regularization to avoid division by zero
    denominator = np.where(np.abs(denominator) < 1e-10, 
                           denominator + 1e-10, 
                           denominator)
    
    Sigma = numerator / denominator
    
    return Sigma

def green_function_down_spin(omega, k, W, J, wm, gamma, s=0.5, eta=1e-4):
    """
    Calculate down-spin Green's function G↓(ω,k) from Eq. (27):
    
    G↓(ω,k) = 1 / [ω - ε_k - Js/2 - Σ(k,ω)]
    
    where ε_k = W k² (parabolic dispersion)
    
    Parameters:
    -----------
    omega : float or array
        Frequency
    k : float
        Electron momentum
    W : float
        Bandwidth parameter
    J : float
        Coupling constant (can be positive or negative)
    wm : float
        Max magnon energy parameter
    gamma : float
        Anisotropy parameter (0 to 1)
    s : float
        Localized spin quantum number
    eta : float
        Small imaginary part for causality
    """
    z = omega + 1j * eta
    
    # Parabolic electron dispersion
    epsilon_k = W * k**2
    
    # Calculate self-energy
    Sigma = self_energy(omega, k, W, J, wm, gamma, s, eta)
    
    # Green's function: G = 1 / [z - ε_k - Js/2 - Σ]
    G_down = 1.0 / (z - epsilon_k - J*s/2 - Sigma)
    
    return G_down

def spectral_weight(omega, k, W, J, wm, gamma, s=0.5, eta=1e-4):
    """
    Calculate spectral weight: ρ↓(k,ω) = -(1/π) Im[G↓(ω,k)]
    
    This is Eq. (44) from the paper.
    """
    G = green_function_down_spin(omega, k, W, J, wm, gamma, s, eta)
    return -np.imag(G) / np.pi

def reproduce_fig2_J_positive(J_mag=1.0, W=10.0, wm=0.01, s=7/2, gamma=1.0):
    """
    Reproduce Fig 2: Down-spin spectral weight for J>0, k=0
    
    Parameters match paper: J=0.2eV, W=2eV normalized to |J|=1
    giving W=10, wm=0.01, s=7/2
    """
    # Set k=0 (band bottom)
    k = 0.0
    
    # For J>0 case (ferromagnetic coupling)
    J = J_mag
    
    # Damping for sharp features
    eta = 0.01
    
    # Omega range
    omega_range = np.linspace(-3, 8, 800)
    
    # Calculate spectral weight
    print("Calculating spectral weight for Fig 2 (J>0, k=0)...")
    rho = np.array([spectral_weight(w, k, W, J, wm, gamma, s, eta) 
                    for w in omega_range])
    
    # Plot vs ω/J + s/2 as in paper
    x_axis = omega_range / J + s/2
    
    # Create figure
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(x_axis, rho, 'b-', linewidth=2, label='Exact')
    
    # Mark Hartree position
    # Hartree: ω = ε_k + Js/2, at k=0: ε_0 = 0, so ω_Hartree = Js/2
    hartree_omega = J * s / 2
    hartree_x = hartree_omega / J + s/2
    ax.axvline(hartree_x, color='r', linestyle='--', linewidth=1.5, 
               label=f'Hartree (ω/J+s/2={hartree_x:.2f})')
    
    ax.set_xlabel(r'$\omega/J + s/2$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Down-spin spectral weight: J>0, k=0\n(γ={gamma}, s={s}, W={W})', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_xlim([x_axis[0], x_axis[-1]])
    ax.set_ylim(bottom=0)
    
    plt.tight_layout()
    return fig

def reproduce_fig3_J_positive_varied_k(J_mag=1.0, W=10.0, wm=0.01, s=7/2, gamma=1.0):
    """
    Reproduce Fig 3: Down-spin spectral weight for J>0 at various ε_k
    
    Plot for ε_k = 0, J, 3J
    For parabolic dispersion ε_k = W k², we need:
    k = sqrt(ε_k / W)
    """
    J = J_mag
    eta = 0.03
    
    # Different band energies: ε_k = 0, J, 3J
    epsilon_values = [0, J_mag, 3*J_mag]
    k_values = [np.sqrt(max(0, eps / W)) for eps in epsilon_values]
    
    fig, ax = plt.subplots(figsize=(10, 6))
    
    colors = ['blue', 'green', 'red']
    
    for idx, (eps_k, k_val) in enumerate(zip(epsilon_values, k_values)):
        omega_range = np.linspace(-2, 10, 600)
        
        print(f"Calculating for ε_k = {eps_k/J_mag:.1f}J, k = {k_val:.3f}...")
        rho = np.array([spectral_weight(w, k_val, W, J, wm, gamma, s, eta) 
                        for w in omega_range])
        
        x_axis = omega_range / J
        label = f'$\\varepsilon_k = {eps_k/J_mag:.0f}J$'
        ax.plot(x_axis, rho, color=colors[idx], linewidth=2, label=label)
        
        # Hartree position for this ε_k
        hartree_omega = eps_k + J * s / 2
        hartree_x = hartree_omega / J
        ax.axvline(hartree_x, color=colors[idx], linestyle='--', 
                   linewidth=1, alpha=0.5)
    
    ax.set_xlabel(r'$\omega/J$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Down-spin spectral weight: J>0, varying $\\varepsilon_k$\n(γ={gamma}, s={s}, W={W})', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim(bottom=0)
    
    plt.tight_layout()
    return fig

def reproduce_fig5_J_negative(J_mag=1.0, W=10.0, wm=0.01, s=7/2, gamma=1.0):
    """
    Reproduce Fig 5: Down-spin spectral weight for J<0, k=0
    Polaronic bound state case
    
    For J<0, bound state appears below continuum
    """
    k = 0.0
    
    # For J<0 case (antiferromagnetic coupling)
    J = -J_mag
    
    # Very small damping to show sharp bound state
    eta = 0.005
    
    omega_range = np.linspace(-3, 4, 1500)
    
    print("Calculating spectral weight for Fig 5 (J<0, k=0, polaron)...")
    rho = np.array([spectral_weight(w, k, W, J, wm, gamma, s, eta) 
                    for w in omega_range])
    
    x_axis = omega_range / J_mag  # Plot vs ω/|J|
    
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(x_axis, rho, 'b-', linewidth=2, label='Exact (bound state)')
    
    # Hartree position 
    # Note: J is negative, so Js/2 is negative
    hartree_omega = J * s / 2  
    hartree_x = hartree_omega / J_mag
    ax.axvline(hartree_x, color='r', linestyle='--', linewidth=1.5, 
               label=f'Hartree (ω/|J|={hartree_x:.2f})')
    
    ax.set_xlabel(r'$\omega/|J|$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Down-spin spectral weight: J<0 (polaron), k=0\n(γ={gamma}, s={s}, W={W})', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim(bottom=0)
    
    # Find and annotate bound state peak
    peak_idx = np.argmax(rho)
    peak_x = x_axis[peak_idx]
    peak_y = rho[peak_idx]
    
    ax.annotate('Bound state\n(polaron)', 
                xy=(peak_x, peak_y * 0.8),
                xytext=(peak_x + 1.5, peak_y * 0.6),
                arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
                fontsize=10, color='red')
    
    plt.tight_layout()
    return fig

def compare_all_figures():
    """
    Generate all three comparison figures matching Shastry-Mattis (1981)
    
    Parameters from paper (Eq. 48 on page 5344):
    J = 0.2 eV, W = 2 eV, ω_m = 0.002 eV, s = 7/2
    
    Normalized to |J| = 1:
    W = 10, ω_m = 0.01, s = 7/2
    """
    print("="*60)
    print("Reproducing Shastry-Mattis (1981) Figures")
    print("Using parabolic dispersion: ε_k = W k²")
    print("Parameters: |J|=1, W=10, ω_m=0.01, s=7/2, γ=1")
    print("="*60)
    print()
    
    # Use paper's parameters
    J_mag = 1.0
    W = 10.0
    wm = 0.01
    s = 7.0/2.0  # s = 7/2 as in paper
    gamma = 1.0
    
    print("Generating Figure 2 (J>0, k=0)...")
    fig2 = reproduce_fig2_J_positive(J_mag, W, wm, s, gamma)
    fig2.savefig('fig2_J_positive_k0.pdf', dpi=300, bbox_inches='tight')
    fig2.savefig('fig2_J_positive_k0.png', dpi=300, bbox_inches='tight')
    print("✓ Saved fig2_J_positive_k0.pdf/png")
    print()
    
    print("Generating Figure 3 (J>0, varied ε_k)...")
    fig3 = reproduce_fig3_J_positive_varied_k(J_mag, W, wm, s, gamma)
    fig3.savefig('fig3_J_positive_varied_k.pdf', dpi=300, bbox_inches='tight')
    fig3.savefig('fig3_J_positive_varied_k.png', dpi=300, bbox_inches='tight')
    print("✓ Saved fig3_J_positive_varied_k.pdf/png")
    print()
    
    print("Generating Figure 5 (J<0, polaron)...")
    fig5 = reproduce_fig5_J_negative(J_mag, W, wm, s, gamma)
    fig5.savefig('fig5_J_negative_polaron.pdf', dpi=300, bbox_inches='tight')
    fig5.savefig('fig5_J_negative_polaron.png', dpi=300, bbox_inches='tight')
    print("✓ Saved fig5_J_negative_polaron.pdf/png")
    print()
    
    print("="*60)
    print("All figures generated successfully!")
    print("="*60)
    
    plt.show()

if __name__ == "__main__":
    compare_all_figures()
```

```python
"""
CORRECTED Shastry-Mattis (1981) Reproduction
===============================================

THREE CRITICAL FIXES:
1. Plot vs (ω - ε_k)/J + s/2, NOT ω/J + s/2
2. Use paper's parameter regime: W/|J|=10, ω_m/W=0.01
3. Enforce retarded boundary condition |ρ| < 1

Key formulas:
- G↓(z,K) = 1 / [z - ε_k + Js/2 - Σ]
- Natural x-axis: (ω - ε_k)/J + s/2
- Hartree appears at x = s (NOT at x = 2s!)
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams

rcParams['font.size'] = 11
rcParams['font.family'] = 'serif'
rcParams['axes.labelsize'] = 11
rcParams['xtick.labelsize'] = 10
rcParams['ytick.labelsize'] = 10

def choose_retarded_rho(alpha, A_plus, A_minus):
    """
    Choose the retarded root with |ρ| < 1 from:
    A_- ρ² + α ρ + A_+ = 0
    """
    disc = alpha**2 - 4.0 * A_plus * A_minus
    sqrt_disc = np.sqrt(disc)
    
    rho1 = (-alpha + sqrt_disc) / (2.0 * A_minus)
    rho2 = (-alpha - sqrt_disc) / (2.0 * A_minus)
    
    # Choose root with smaller magnitude (retarded condition)
    abs_rho1 = np.abs(rho1)
    abs_rho2 = np.abs(rho2)
    
    # Use the one with |ρ| < 1, or if both fail, the smaller one
    if abs_rho1 < 1.0 and abs_rho1 <= abs_rho2:
        return rho1
    elif abs_rho2 < 1.0:
        return rho2
    else:
        # Edge case: both ≥ 1, pick smaller
        return rho1 if abs_rho1 < abs_rho2 else rho2

def green_function_down_spin(omega, K, t, J0, J1, gamma, s=7/2, eta=1e-3):
    """
    Down-spin Green's function with CORRECT conventions:
    
    G↓(z,K) = 1 / [z - ε_k + Js/2 - Σ]
    
    where:
    - ε_k = -2t cos(K) (tight-binding)
    - Σ = (γ²J₁²/4) / (β + 2A_-ρ)
    - α = z - J₀ - Js/2
    - β = z - J₀ + Js/2
    
    Parameters:
    -----------
    omega : float
        Frequency
    K : float
        Crystal momentum
    t : float
        Electron hopping (W ≈ 4t for 1D tight-binding)
    J0 : float
        Magnon coupling (ω_m = 2J₀)
    J1 : float
        Electron-magnon coupling (signed: >0 FM, <0 AFM)
    gamma : float
        Anisotropy (0=Ising, 1=Heisenberg)
    s : float
        Localized spin quantum number
    eta : float
        Damping for retarded Green's function
    """
    z = omega + 1j * eta
    
    # Electron dispersion
    epsilon_k = -2.0 * t * np.cos(K)
    
    # A_± parameters
    A_plus = t * np.exp(1j * K) + J0 / 2.0
    A_minus = t * np.exp(-1j * K) + J0 / 2.0
    
    # α, β with general s
    alpha = z - J0 - J1 * s / 2.0
    beta = z - J0 + J1 * s / 2.0
    
    # Retarded root with |ρ| < 1
    rho = choose_retarded_rho(alpha, A_plus, A_minus)
    
    # Self-energy
    denom = beta + 2.0 * A_minus * rho
    # Tiny regularization if needed
    if np.abs(denom) < 1e-14:
        denom = denom + 1e-14j
    
    Sigma = (gamma**2 * J1**2 / 4.0) / denom
    
    # Green's function: G↓ = 1 / [z - ε_k + Js/2 - Σ]
    G_down = 1.0 / (z - epsilon_k + J1 * s / 2.0 - Sigma)
    
    return G_down

def spectral_weight(omega, K, t, J0, J1, gamma, s=7/2, eta=1e-3):
    """
    Spectral weight: ρ↓(k,ω) = -(1/π) Im[G↓]
    """
    G = green_function_down_spin(omega, K, t, J0, J1, gamma, s, eta)
    return -np.imag(G) / np.pi

def reproduce_fig2_J_positive_CORRECTED(J1=1.0, s=7/2, gamma=1.0):
    """
    Fig 2: J>0, k=0, ε_k=0
    
    CORRECTED parameters matching paper:
    - W = 10|J| → t = 2.5 (since W ≈ 4t)
    - ω_m/W = 0.01 → ω_m = 0.1 → J₀ = 0.05
    - Plot vs (ω - ε_k)/J + s/2
    - Hartree appears at x = s = 3.5
    """
    K = 0.0
    t = 2.5  # W ≈ 10
    J0 = 0.05  # ω_m = 0.1
    eta = 0.01
    
    print(f"Fig 2: J1={J1}, K={K}, t={t}, J0={J0}, s={s}")
    print(f"       W≈{4*t:.1f}, ω_m={2*J0:.2f}, ω_m/W={2*J0/(4*t):.3f}")
    
    # Electron dispersion at K=0
    epsilon_k = -2.0 * t * np.cos(K)
    print(f"       ε_k = {epsilon_k:.3f}")
    
    # Omega range (wide enough to see all features)
    omega_range = np.linspace(-2, 8, 2000)
    
    rho = np.array([spectral_weight(w, K, t, J0, J1, gamma, s, eta) 
                    for w in omega_range])
    
    # CRITICAL FIX: Plot vs (ω - ε_k)/J + s/2
    x_axis = (omega_range - epsilon_k) / J1 + s/2
    
    # Hartree position
    hartree_omega = epsilon_k + J1 * s / 2
    hartree_x = (hartree_omega - epsilon_k) / J1 + s/2
    print(f"       Hartree: ω={hartree_omega:.3f}, x={hartree_x:.3f}")
    
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(x_axis, rho, 'b-', linewidth=2, label='Exact')
    ax.axvline(hartree_x, color='r', linestyle='--', linewidth=1.5, 
               label=f'Hartree (x={hartree_x:.1f})')
    
    ax.set_xlabel(r'$(\omega - \varepsilon_k)/J + s/2$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Fig 2: J>0, k=0 (s={s}, γ={gamma})\n' +
                 f'W/|J|={4*t:.0f}, ω_m/W={2*J0/(4*t):.3f}', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_xlim([1, 6])
    ax.set_ylim(bottom=0)
    
    plt.tight_layout()
    return fig

def reproduce_fig3_J_positive_varied_CORRECTED(J1=1.0, s=7/2, gamma=1.0):
    """
    Fig 3: J>0, varying ε_k = 0, J, 3J
    
    For tight-binding with t=2.5:
    - ε_k = 0: choose K such that -2t cos(K) = 0 → K = π/2
    - ε_k = J: choose K such that -2t cos(K) = J → cos(K) = -J/(2t)
    - ε_k = 3J: similar
    """
    t = 2.5
    J0 = 0.05
    eta = 0.03
    
    print(f"Fig 3: Varying ε_k with t={t}, J0={J0}")
    
    # Target ε_k values
    epsilon_targets = [0.0, J1, 3.0 * J1]
    colors = ['blue', 'green', 'red']
    
    fig, ax = plt.subplots(figsize=(10, 6))
    
    for idx, eps_target in enumerate(epsilon_targets):
        # Find K that gives ε_k = eps_target
        # ε_k = -2t cos(K) = eps_target
        # cos(K) = -eps_target / (2t)
        cos_K = -eps_target / (2.0 * t)
        
        if np.abs(cos_K) <= 1.0:
            K = np.arccos(cos_K)
        else:
            # Out of range, use K=0 or K=π
            K = 0.0 if eps_target < 0 else np.pi
        
        epsilon_k = -2.0 * t * np.cos(K)
        print(f"  ε_k={eps_target:.2f}: K={K:.3f}, actual ε_k={epsilon_k:.3f}")
        
        omega_range = np.linspace(-2, 12, 1500)
        rho = np.array([spectral_weight(w, K, t, J0, J1, gamma, s, eta) 
                        for w in omega_range])
        
        # Plot vs (ω - ε_k)/J + s/2
        x_axis = (omega_range - epsilon_k) / J1 + s/2
        
        label = f'$\\varepsilon_k = {eps_target/J1:.0f}J$'
        ax.plot(x_axis, rho, color=colors[idx], linewidth=2, label=label)
        
        # Hartree line
        hartree_x = s  # Always at x = s in shifted coordinates!
        ax.axvline(hartree_x, color=colors[idx], linestyle='--', 
                   linewidth=1, alpha=0.5)
    
    ax.set_xlabel(r'$(\omega - \varepsilon_k)/J + s/2$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Fig 3: J>0, varying $\\varepsilon_k$ (s={s}, γ={gamma})\n' +
                 f'W/|J|={4*t:.0f}, ω_m/W={2*J0/(4*t):.3f}', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_xlim([1, 6])
    ax.set_ylim(bottom=0)
    
    plt.tight_layout()
    return fig

def reproduce_fig5_J_negative_CORRECTED(J1=-1.0, s=7/2, gamma=1.0):
    """
    Fig 5: J<0, k=0, polaron bound state
    
    Bound state appears BELOW Hartree at x < s
    """
    K = 0.0
    t = 2.5
    J0 = -0.05  # Same magnitude, opposite sign
    eta = 0.005  # Small for sharp bound state
    
    print(f"Fig 5: J1={J1}, K={K}, t={t}, J0={J0}, s={s}")
    
    epsilon_k = -2.0 * t * np.cos(K)
    print(f"       ε_k = {epsilon_k:.3f}")
    
    omega_range = np.linspace(-8, 4, 3000)
    rho = np.array([spectral_weight(w, K, t, J0, J1, gamma, s, eta) 
                    for w in omega_range])
    
    # Plot vs (ω - ε_k)/|J| + s/2
    x_axis = (omega_range - epsilon_k) / np.abs(J1) + s/2
    
    # Hartree (should be above bound state)
    hartree_omega = epsilon_k + J1 * s / 2  # J1 < 0, so this is LOWER
    hartree_x = (hartree_omega - epsilon_k) / np.abs(J1) + s/2
    print(f"       Hartree: x={hartree_x:.3f}")
    
    # Find bound state peak
    peak_idx = np.argmax(rho)
    peak_x = x_axis[peak_idx]
    peak_omega = omega_range[peak_idx]
    binding_energy = hartree_omega - peak_omega
    print(f"       Bound state: x={peak_x:.3f}, ω={peak_omega:.3f}")
    print(f"       Binding energy: {binding_energy:.3f}")
    
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(x_axis, rho, 'b-', linewidth=2, label='Exact (bound state)')
    ax.axvline(hartree_x, color='r', linestyle='--', linewidth=1.5, 
               label=f'Hartree (x={hartree_x:.1f})')
    ax.axvline(peak_x, color='orange', linestyle=':', linewidth=1.5,
               label=f'Bound state (x={peak_x:.1f})')
    
    ax.set_xlabel(r'$(\omega - \varepsilon_k)/|J| + s/2$', fontsize=12)
    ax.set_ylabel(r'$\rho_\downarrow(k,\omega)$', fontsize=12)
    ax.set_title(f'Fig 5: J<0, k=0, polaron (s={s}, γ={gamma})\n' +
                 f'W/|J|={4*t:.0f}, ω_m/W={2*np.abs(J0)/(4*t):.3f}', 
                 fontsize=13)
    ax.legend()
    ax.grid(True, alpha=0.3)
    ax.set_ylim(bottom=0)
    
    # Annotate
    if rho[peak_idx] > 0.5:
        ax.annotate(f'E_b ≈ {binding_energy:.2f}', 
                    xy=(peak_x, rho[peak_idx] * 0.7),
                    xytext=(peak_x - 1, rho[peak_idx] * 0.5),
                    arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
                    fontsize=10, color='red')
    
    plt.tight_layout()
    return fig

def main():
    """
    Generate ALL THREE figures with CORRECT parameters and axis
    """
    print("="*70)
    print("CORRECTED Shastry-Mattis (1981) Reproduction")
    print("="*70)
    print("\nKey fixes:")
    print("1. Plot vs (ω - ε_k)/J + s/2")
    print("2. Use W/|J|=10, ω_m/W=0.01")
    print("3. Enforce |ρ| < 1")
    print("="*70)
    print()
    
    # Figure 2: J>0, k=0
    print("\n" + "="*70)
    print("FIGURE 2: J>0, k=0")
    print("="*70)
    fig2 = reproduce_fig2_J_positive_CORRECTED()
    fig2.savefig('fig2_CORRECTED.pdf', dpi=300, bbox_inches='tight')
    fig2.savefig('fig2_CORRECTED.png', dpi=300, bbox_inches='tight')
    print("✓ Saved fig2_CORRECTED.pdf/png")
    
    # Figure 3: J>0, varied ε_k
    print("\n" + "="*70)
    print("FIGURE 3: J>0, varying ε_k")
    print("="*70)
    fig3 = reproduce_fig3_J_positive_varied_CORRECTED()
    fig3.savefig('fig3_CORRECTED.pdf', dpi=300, bbox_inches='tight')
    fig3.savefig('fig3_CORRECTED.png', dpi=300, bbox_inches='tight')
    print("✓ Saved fig3_CORRECTED.pdf/png")
    
    # Figure 5: J<0, polaron
    print("\n" + "="*70)
    print("FIGURE 5: J<0, polaron")
    print("="*70)
    fig5 = reproduce_fig5_J_negative_CORRECTED()
    fig5.savefig('fig5_CORRECTED.pdf', dpi=300, bbox_inches='tight')
    fig5.savefig('fig5_CORRECTED.png', dpi=300, bbox_inches='tight')
    print("✓ Saved fig5_CORRECTED.pdf/png")
    
    print("\n" + "="*70)
    print("ALL FIGURES GENERATED SUCCESSFULLY!")
    print("="*70)
    print("\nExpected features:")
    print("  Fig 2: Single broad peak at x ≈ 3.5 (Hartree)")
    print("  Fig 3: Three peaks all near x ≈ 3.5, width increases")
    print("  Fig 5: Sharp bound state BELOW Hartree (x < 3.5)")
    print("="*70)
    
    plt.show()

if __name__ == "__main__":
    main()
```

```python
"""
Shastry-Mattis Magnetic Polaron Theory
Reproduces Figures 2, 3, 5 from Phys. Rev. B 24, 5340 (1981)

Theory of a single electron coupled to ferromagnetic spin background
with spin-flip processes and magnon emission/absorption.
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

class MagneticPolaron:
    """
    Exact solution for magnetic polaron problem.
    
    Parameters from paper (in units of |J|=1):
    - J = ±1 (ferromagnetic +1, antiferromagnetic -1)
    - W = 10 (bandwidth)
    - w_m = 0.01 (max magnon energy)
    - s = 7/2 (localized spin)
    - gamma = 1 (Heisenberg limit)
    """
    
    def __init__(self, J, W=10.0, w_m=0.01, s=3.5, gamma=1.0):
        self.J = J
        self.W = W
        self.w_m = w_m
        self.s = s
        self.gamma = gamma
        
    def epsilon_k(self, k):
        """Parabolic electron band: ε_k = W k²"""
        return self.W * k**2
    
    def omega_q(self, q):
        """Parabolic magnon: ω_q = w_m q²"""
        return self.w_m * q**2
    
    def v_k(self, k, omega):
        """
        Shifted energy variable from Eq. (35):
        v_k = ω + Js/2 - (w_m/(w_m+W))ε_k
        """
        eps_k = self.epsilon_k(k)
        return omega + self.J*self.s/2.0 - (self.w_m/(self.w_m + self.W))*eps_k
    
    def compute_A(self, k, omega):
        """
        Compute A(k,ω) = A₁ + i A₂ from Eqs. (37-40)
        
        This is the key function that enters the self-energy.
        """
        v = self.v_k(k, omega)
        W_plus_wm = self.W + self.w_m
        
        if v < 0:
            # v_k < 0: Eqs. (39-40)
            abs_v = abs(v)
            sqrt_ratio = np.sqrt(abs_v / W_plus_wm)
            
            A1 = (3.0 / W_plus_wm) * (1.0 - sqrt_ratio * np.arctan(1.0/sqrt_ratio))
            A2 = 0.0  # θ(v_k) = 0 for v < 0
            
        else:
            # v_k > 0: Eqs. (40)
            sqrt_v = np.sqrt(v)
            sqrt_W = np.sqrt(W_plus_wm)
            
            # Handle log singularity carefully
            if v < W_plus_wm * 0.9999:  # avoid denominator → 0
                ratio = (sqrt_W + sqrt_v) / (sqrt_W - sqrt_v)
                A1 = (3.0 / W_plus_wm) * (1.0 - 0.5 * np.sqrt(v/W_plus_wm) * np.log(abs(ratio)))
            else:
                # Near singularity, use asymptotic form
                A1 = (3.0 / W_plus_wm) * (1.0 - 0.5 * np.sqrt(v/W_plus_wm) * 10.0)  # large value
            
            # Imaginary part from Eq. (38)
            A2 = (3.0 * np.pi / (2.0 * W_plus_wm**(3/2))) * sqrt_v
            
        return A1, A2
    
    def compute_self_energy(self, k, omega):
        """
        Compute self-energy Σ(k,ω) = Σ₁ + i Σ₂ from Eqs. (42-43)
        
        Σ₁ = Js γ² [(1 + J/2 A₁) / ((1 + J/2 A₁)² + (J²/4)A₂²) - 1]
        
        Σ₂ = -(J²s/2) γ² A₂ / ((1 + J/2 A₁)² + (J²/4)A₂²)
        """
        A1, A2 = self.compute_A(k, omega)
        
        # Common denominator
        term1 = 1.0 + (self.J/2.0)*A1
        term2 = (self.J**2 / 4.0) * A2**2
        denom = term1**2 + term2
        
        # Real part of self-energy
        Sigma1 = self.J * self.s * self.gamma**2 * (term1 / denom - 1.0)
        
        # Imaginary part of self-energy
        Sigma2 = -(self.J**2 * self.s / 2.0) * self.gamma**2 * (A2 / denom)
        
        return Sigma1, Sigma2
    
    def spectral_weight(self, k, omega, eta=1e-4):
        """
        Down-spin spectral weight from Eq. (45):
        
        ρ↓(k,ω) = (1/π) * η / [(η - Σ₂)² + (ω - ε_k - Js/2 - Σ₁)²]
        
        where η = 0⁺ is small positive number (broadening)
        """
        eps_k = self.epsilon_k(k)
        Sigma1, Sigma2 = self.compute_self_energy(k, omega)
        
        # Hartree energy
        E_hartree = eps_k + self.J * self.s / 2.0
        
        # Spectral function
        numerator = eta - Sigma2  # This should be positive
        denominator = (eta - Sigma2)**2 + (omega - E_hartree - Sigma1)**2
        
        rho = (1.0 / np.pi) * abs(numerator) / denominator
        
        return rho
    
    def density_of_states(self, omega, k_max=1.0, nk=100, eta=1e-4):
        """
        Total density of states: ρ(ω) = (1/N) Σ_k ρ(k,ω)
        
        Integrate over k in spherical approximation.
        """
        k_values = np.linspace(0, k_max, nk)
        
        dos = 0.0
        for k in k_values:
            if k == 0:
                weight = 0  # No states at k=0 in 3D
            else:
                weight = k**2  # 3D density of states ∝ k²
            dos += weight * self.spectral_weight(k, omega, eta)
        
        # Normalize
        dos = dos / np.sum(k_values**2) if np.sum(k_values**2) > 0 else dos
        
        return dos


def plot_figure_2():
    """
    Reproduce Figure 2: Spectral weight for k=0, J>0
    Shows the broad scattering peak near Hartree value and weak bound state above.
    """
    print("Generating Figure 2: J > 0, k = 0...")
    
    polaron = MagneticPolaron(J=+1.0, W=10.0, w_m=0.01, s=3.5, gamma=1.0)
    
    # Energy range
    # Hartree energy at k=0: E = 0 + J*s/2 = 1.75
    omega_values = np.linspace(-2.0, 10.0, 1000)
    
    k = 0.0
    rho_values = np.array([polaron.spectral_weight(k, omega, eta=0.02) for omega in omega_values])
    
    # Hartree prediction
    E_hartree = 0.0 + polaron.J * polaron.s / 2.0
    
    fig, ax = plt.subplots(figsize=(6, 6))
    
    ax.plot(omega_values, rho_values, 'b-', linewidth=2, label='Exact')
    ax.axvline(E_hartree, color='red', linestyle='--', linewidth=2, label=f'Hartree: ω = {E_hartree:.2f}')
    
    # Mark bound state (should be above scattering continuum for J>0)
    # The bound state is where G⁻¹ = 0, which is hard to locate analytically
    # but should appear as a peak above the main continuum
    
    ax.set_xlabel('ω/J+s/2', fontsize=14)
    ax.set_ylabel('ρ↓(k=0, ω)', fontsize=14)
    ax.set_title('Down-Spin Spectral Weight: J > 0, k = 0', fontsize=16)
    ax.legend(fontsize=12)
    ax.grid(True, alpha=0.3)
    ax.set_xlim(-2, 10)
    
    plt.tight_layout()
    plt.savefig('shastry_mattis_fig2.png', dpi=150, bbox_inches='tight')
    print("Saved: shastry_mattis_fig2.png")
    
    return fig


def plot_figure_3():
    """
    Reproduce Figure 3: Spectral weight for J>0 at ε_k = 0, J, 3J
    Shows how linewidth increases with increasing k (more phase space for magnon emission)
    """
    print("Generating Figure 3: J > 0, multiple k values...")
    
    polaron = MagneticPolaron(J=+1.0, W=10.0, w_m=0.01, s=3.5, gamma=1.0)
    
    # For ε_k = Wk², we need k = sqrt(ε_k/W)
    epsilon_values = [0.0, 1.0, 3.0]  # ε_k = 0, J, 3J
    k_values = [np.sqrt(eps/polaron.W) for eps in epsilon_values]
    
    omega_values = np.linspace(-2.0, 10.0, 1000)
    
    fig, ax = plt.subplots(figsize=(6, 6))
    
    colors = ['blue', 'green', 'red']
    for i, (k, eps) in enumerate(zip(k_values, epsilon_values)):
        rho_values = np.array([polaron.spectral_weight(k, omega, eta=0.02) for omega in omega_values])
        
        E_hartree = eps + polaron.J * polaron.s / 2.0
        
        ax.plot(omega_values, rho_values, color=colors[i], linewidth=2, 
                label=f'ε_k = {eps:.1f}J')
        ax.axvline(E_hartree, color=colors[i], linestyle='--', linewidth=1, alpha=0.5)
    
    ax.set_xlabel('ω/J', fontsize=14)
    ax.set_ylabel('ρ↓(k, ω)', fontsize=14)
    ax.set_title('Down-Spin Spectral Weight: J > 0, Various k', fontsize=16)
    ax.legend(fontsize=12)
    ax.grid(True, alpha=0.3)
    ax.set_xlim(-2, 10)
    
    plt.tight_layout()
    plt.savefig('shastry_mattis_fig3.png', dpi=150, bbox_inches='tight')
    print("Saved: shastry_mattis_fig3.png")
    
    return fig


def plot_figure_5():
    """
    Reproduce Figure 5: Spectral weight for k=0, J<0 (antiferromagnetic)
    Shows sharp bound state (polaron) below scattering continuum.
    """
    print("Generating Figure 5: J < 0, k = 0 (polaron state)...")
    
    polaron = MagneticPolaron(J=-1.0, W=10.0, w_m=0.01, s=3.5, gamma=1.0)
    
    # Energy range
    # Hartree energy at k=0: E = 0 - J*s/2 = -1.75
    omega_values = np.linspace(-4.0, 5.0, 1000)
    
    k = 0.0
    rho_values = np.array([polaron.spectral_weight(k, omega, eta=0.005) for omega in omega_values])
    
    # Hartree prediction
    E_hartree = 0.0 + polaron.J * polaron.s / 2.0  # = -1.75
    
    fig, ax = plt.subplots(figsize=(6, 6))
    
    ax.plot(omega_values, rho_values, 'b-', linewidth=2, label='Exact (Polaron)')
    ax.axvline(E_hartree, color='red', linestyle='--', linewidth=2, 
               label=f'Hartree: ω = {E_hartree:.2f}')
    
    # The bound state should be at lower energy than Hartree (binding energy ≈ 1.5 J²/W from Eq. 49)
    binding_energy = 1.5 * polaron.J**2 / polaron.W
    E_bound_estimate = E_hartree - binding_energy
    ax.axvline(E_bound_estimate, color='green', linestyle=':', linewidth=2,
               label=f'Bound state est: ω ≈ {E_bound_estimate:.2f}')
    
    ax.set_xlabel('ω/J', fontsize=14)
    ax.set_ylabel('ρ↓(k=0, ω)', fontsize=14)
    ax.set_title('Down-Spin Spectral Weight: J < 0 (Polaron), k = 0', fontsize=16)
    ax.legend(fontsize=12)
    ax.grid(True, alpha=0.3)
    ax.set_xlim(-4, 5)
    
    
    plt.tight_layout()
    plt.savefig('shastry_mattis_fig5.png', dpi=150, bbox_inches='tight')
    print("Saved: shastry_mattis_fig5.png")
    
    return fig


def plot_combined_comparison():
    """
    Create a combined figure showing all three cases side by side.
    """
    print("Generating combined comparison figure...")
    
    fig = plt.figure(figsize=(15, 5))
    gs = GridSpec(1, 3, figure=fig)
    
    # Figure 2: J > 0, k = 0
    ax1 = fig.add_subplot(gs[0, 0])
    polaron_ferro = MagneticPolaron(J=+1.0, W=10.0, w_m=0.01, s=3.5, gamma=1.0)
    omega_vals = np.linspace(-2, 10, 500)
    rho_ferro = [polaron_ferro.spectral_weight(0.0, omega, eta=0.02) for omega in omega_vals]
    ax1.plot(omega_vals, rho_ferro, 'b-', linewidth=2)
    ax1.axvline(polaron_ferro.J * polaron_ferro.s / 2, color='r', linestyle='--', label='Hartree')
    ax1.set_xlabel('ω/J+s/2', fontsize=12)
    ax1.set_ylabel('ρ↓(k, ω)', fontsize=12)
    ax1.set_title('(a) J > 0, k = 0\n(Ferromagnetic)', fontsize=12)
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Figure 3: J > 0, multiple k
    ax2 = fig.add_subplot(gs[0, 1])
    epsilon_values = [0.0, 1.0, 3.0]
    colors = ['blue', 'green', 'red']
    for eps, color in zip(epsilon_values, colors):
        k = np.sqrt(eps/polaron_ferro.W)
        rho_vals = [polaron_ferro.spectral_weight(k, omega, eta=0.02) for omega in omega_vals]
        ax2.plot(omega_vals, rho_vals, color=color, linewidth=2, label=f'ε={eps:.1f}J')
    ax2.set_xlabel('ω/J', fontsize=12)
    ax2.set_ylabel('ρ↓(k, ω)', fontsize=12)
    ax2.set_title('(b) J > 0, various k\n(Linewidth increases with k)', fontsize=12)
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # Figure 5: J < 0, k = 0
    ax3 = fig.add_subplot(gs[0, 2])
    polaron_afm = MagneticPolaron(J=-1.0, W=10.0, w_m=0.01, s=3.5, gamma=1.0)
    omega_vals_afm = np.linspace(-4, 5, 500)
    rho_afm = [polaron_afm.spectral_weight(0.0, omega, eta=0.005) for omega in omega_vals_afm]
    ax3.plot(omega_vals_afm, rho_afm, 'b-', linewidth=2)
    ax3.axvline(polaron_afm.J * polaron_afm.s / 2, color='r', linestyle='--', label='Hartree')
    ax3.set_xlabel('ω/J', fontsize=12)
    ax3.set_ylabel('ρ↓(k, ω)', fontsize=12)
    ax3.set_title('(c) J < 0, k = 0\n(Antiferromagnetic Polaron)', fontsize=12)
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('shastry_mattis_combined.png', dpi=150, bbox_inches='tight')
    print("Saved: shastry_mattis_combined.png")
    
    return fig


def analyze_physics():
    """
    Print key physical insights from the theory.
    """
    print("\n" + "="*70)
    print("PHYSICAL INTERPRETATION OF SHASTRY-MATTIS MAGNETIC POLARON")
    print("="*70)
    
    print("\n1. FERROMAGNETIC COUPLING (J > 0):")
    print("   - Spin-down electron is UNSTABLE")
    print("   - Decay time: τ ~ W/(γ²J²)")
    print("   - Process: e↓ → e↑ + magnon")
    print("   - Spectral weight dominated by SCATTERING STATES")
    print("   - Electron effective repels magnon")
    print("   - Bound state above continuum (negligible weight)")
    
    polaron_ferro = MagneticPolaron(J=1.0, W=10.0)
    tau_ferro = polaron_ferro.W / (polaron_ferro.gamma**2 * polaron_ferro.J**2)
    distance_ferro = np.sqrt(tau_ferro)
    print(f"   - Typical decay time: τ ~ {tau_ferro:.1f} (in units of J⁻¹)")
    print(f"   - Distance traveled: ~ {distance_ferro:.0f} lattice sites\n")
    
    print("2. ANTIFERROMAGNETIC COUPLING (J < 0):")
    print("   - Spin-down electron forms STABLE POLARON")
    print("   - Bound state below continuum (dominant weight)")
    print("   - Binding energy: E_B ≈ 1.5 J²/W (weak coupling)")
    print("   - Electron attracts magnon cloud")
    print("   - Electron spin screening: <σ^z> ≈ -1/2 + O(J²/W)")
    
    polaron_afm = MagneticPolaron(J=-1.0, W=10.0)
    E_bind = 1.5 * polaron_afm.J**2 / polaron_afm.W
    print(f"   - Binding energy: E_B ≈ {E_bind:.3f}J")
    print(f"   - Polaron is dressed by magnon cloud\n")
    
    print("3. KEY DIFFERENCE:")
    print("   - SIGN OF J determines qualitatively different physics")
    print("   - Hartree approx FAILS for bulk ferromagnets (J>0)")
    print("   - Hartree approx OK for thin films or antiferromagnets (J<0)")
    print("   - Discontinuity at J=0 (Fig. 1 in paper)\n")
    
    print("4. EXPERIMENTAL RELEVANCE:")
    print("   - EuO, EuS: ferromagnetic semiconductors (J > 0)")
    print("   - Photoemission: sees both e↑ and e↓ despite J > 0")
    print("   - Explanation: spin-down decays rapidly to spin-up + magnon")
    print("   - Lifetime broadening visible in spectra\n")
    
    print("="*70 + "\n")


if __name__ == "__main__":
    print("Shastry-Mattis Magnetic Polaron Theory")
    print("Phys. Rev. B 24, 5340 (1981)\n")
    
    # Print physical interpretation
    analyze_physics()
    
    # Generate all figures
    plot_figure_2()
    plot_figure_3()
    plot_figure_5()
    plot_combined_comparison()
    
    print("\n" + "="*70)
    print("All figures generated successfully!")
    print("="*70)
    print("\nFiles created:")
    print("  - shastry_mattis_fig2.png  (J>0, k=0: scattering states)")
    print("  - shastry_mattis_fig3.png  (J>0, various k: increasing linewidth)")
    print("  - shastry_mattis_fig5.png  (J<0, k=0: polaron bound state)")
    print("  - shastry_mattis_combined.png  (all three side-by-side)")
```

```python
import numpy as np
import matplotlib.pyplot as plt

class RealSpaceMagneticPolaron:
    """
    Real-space magnetic polaron Green's function (one-magnon sector, general s).

    Conventions (match the LaTeX):
      A_± = t e^{±iK} + s J0
      alpha = z - 2 J0 s - (J1 s)/2
      beta  = z - 2 J0 s - J1 (s-1)/2
      Σ(z,K) = (γ^2 J1^2 s / 2) / (beta + 2 A_- ρ)
      G(z,K,-) = 1 / [ z + 2 t cos K + (J1 s)/2 - Σ(z,K) ]

    Note: for ω_q = 2 J0 s (1 - cos q), the magnon bandwidth is ω_max = 4 J0 s,
          so J0 = ω_m / (4 s).
    """
    def __init__(self, t, J_0, J_1, s, gamma=1.0):
        self.t = t
        self.J_0 = J_0
        self.J_1 = J_1
        self.s = s
        self.gamma = gamma

    def epsilon_k(self, K):
        return -2.0 * self.t * np.cos(K)

    def compute_rho(self, z, K):
        A_plus  = self.t * np.exp(1j*K) + self.s * self.J_0
        A_minus = self.t * np.exp(-1j*K) + self.s * self.J_0

        alpha = z - 2.0*self.J_0*self.s - 0.5*self.J_1*self.s
        disc = alpha*alpha - 4.0 * A_plus * A_minus
        sqrt_disc = np.sqrt(disc + 0j)

        rho1 = (-alpha + sqrt_disc) / (2.0 * A_minus)
        rho2 = (-alpha - sqrt_disc) / (2.0 * A_minus)
        rho = rho1 if abs(rho1) < abs(rho2) else rho2
        return rho, A_minus

    def compute_G_minus(self, z, K):
        rho, A_minus = self.compute_rho(z, K)
        beta = z - 2.0*self.J_0*self.s - 0.5*self.J_1*(self.s - 1.0)

        # Correct vertex: g^2 = (γ J1)^2 * s / 2
        vertex_sq = (self.gamma * self.J_1)**2 * self.s / 2.0
        Sigma = vertex_sq / (beta + 2.0 * A_minus * rho)

        # Denominator with ε_k = -2t cosK
        denom = z + 2.0*self.t*np.cos(K) + 0.5*self.J_1*self.s - Sigma
        return 1.0 / denom

    def spectral_weight(self, K, omega, eta=2e-2):
        z = omega + 1j*eta
        Gm = self.compute_G_minus(z, K)
        return -np.imag(Gm) / np.pi

    # Shastry–Mattis x-axis helper; if you keep ε_0 = -2t, subtracting ε_0 is
    # equivalent to shifting the zero so the band bottom is at 0.
    def x_SM(self, omega, K=0.0):
        return (omega)


# ----- Parameters (consistent with SM Fig. 2) -----
s = 0.5
J1 = 5         # choose |J| = 1 as energy unit
W = 10.0
t = 1      # W = 4t
#omega_m = 0.01
J0 = 0.05#omega_m / (4.0 * s)   # correct mapping
gamma = 0.5

pol = RealSpaceMagneticPolaron(t, J0, J1, s, gamma)

K = 0.0

# Work in physical ω; for reference ε_0 = -2t and Hartree ω_H = ε_0 + J1*s/2
omega = np.linspace(-6, +2, 1500)
eta = 0.02
rho = np.array([pol.spectral_weight(K, w, eta=eta) for w in omega])

# SM display axis; subtract ε_0 (same as recentering band bottom to 0)
x = pol.x_SM(omega, K=K)  # = (ω - ε_0)/J1 + s/2

plt.figure(figsize=(7,4))
plt.plot(x, rho, lw=2)
plt.axvline(s/2, ls='--', c='r', label=r'Hartree: $s/2$')
plt.xlabel(r'$ (\omega - \varepsilon_{K})/J + s/2 $')
plt.ylabel(r'$\rho_\downarrow(K{=}0,\omega)$')
plt.ylim(0,1)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar

s = 0.5
J1 = 5         # choose |J| = 1 as energy unit
W = 10.0
t = 1      # W = 4t
#omega_m = 0.01
J0 = 0.05#omega_m / (4.0 * s)   # correct mapping
gamma = 0.5

def energy_function(q, K, t, J0, s):
    """
    Energy as function of q for fixed K:
    E(K,q) = 2t cos(K - q) + 4 J0 s sin^2(q/2)
    
    First term: electron kinetic energy at K-q
    Second term: magnon energy ω_q
    """
    electron_term = 2 * t * np.cos(K - q)
    magnon_term = 4 * J0 * s * np.sin(q/2)**2
    return electron_term + magnon_term

# Create q array for plotting
q_array = np.linspace(-np.pi, np.pi, 1000)

# Test for several K values
k_values = np.linspace(0, np.pi, 200)

# Initialize arrays (THIS WAS MISSING!)
q_min_array = []
E_min_array = []

for K in k_values:
    result = minimize_scalar(lambda q: energy_function(q, K, t, J0, s), 
                            bounds=(-np.pi, np.pi), method='bounded')
    q_min_array.append(result.x)
    E_min_array.append(result.fun)

q_min_array = np.array(q_min_array)
E_min_array = np.array(E_min_array)

plt.plot(K_values, q_min_array + E_min_array)
plt.xlabel('K')
plt.ylabel('q_min + E_min')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

class RealSpaceMagneticPolaron:
    """
    Real-space magnetic polaron Green's function (one-magnon sector, general s).
    Conventions (match the LaTeX):
      A_± = t e^{±iK} + s J0
      alpha = z - 2 J0 s - (J1 s)/2
      beta  = z - 2 J0 s - J1 (s-1)/2
      Σ(z,K) = (γ^2 J1^2 s / 2) / (beta + 2 A_- ρ)
      G(z,K,-) = 1 / [ z + 2 t cos K + (J1 s)/2 - Σ(z,K) ]
    Note: for ω_q = 2 J0 s (1 - cos q), the magnon bandwidth is ω_max = 4 J0 s,
          so J0 = ω_m / (4 s).
    """
    def __init__(self, t, J_0, J_1, s, gamma=1.0):
        self.t = t
        self.J_0 = J_0
        self.J_1 = J_1
        self.s = s
        self.gamma = gamma
    
    def epsilon_k(self, K):
        return 2.0 * self.t * (np.cos(K))
    
    def compute_rho(self, z, K):
        A_plus  = self.t * np.exp(1j*K) + self.s * self.J_0
        A_minus = self.t * np.exp(-1j*K) + self.s * self.J_0
        alpha = z - 2.0*self.J_0*self.s - 0.5*self.J_1*self.s
        disc = alpha*alpha - 4.0 * A_plus * A_minus
        sqrt_disc = np.sqrt(disc + 0j)
        rho1 = (-alpha + sqrt_disc) / (2.0 * A_minus)
        rho2 = (-alpha - sqrt_disc) / (2.0 * A_minus)
        rho = rho1 if abs(rho1) < abs(rho2) else rho2
        return rho, A_minus
    
    def compute_G_minus(self, z, K):
        rho, A_minus = self.compute_rho(z, K)
        beta = z - 2.0*self.J_0*self.s - 0.5*self.J_1*(self.s - 1.0)
        # Correct vertex: g^2 = (γ J1)^2 * s / 2
        vertex_sq = (self.gamma * self.J_1)**2 * self.s / 2.0
        Sigma = vertex_sq / (beta + 2.0 * A_minus * rho)
        # Denominator with ε_k = -2t cosK
        denom = z + 2.0*self.t*np.cos(K) + 0.5*self.J_1*self.s - Sigma
        return 1.0 / denom
    
    def spectral_weight(self, K, omega, eta=2e-2):
        z = omega + 1j*eta
        Gm = self.compute_G_minus(z, K)
        return -np.imag(Gm) / np.pi
    
    # Shastry–Mattis x-axis helper; if you keep ε_0 = -2t, subtracting ε_0 is
    # equivalent to shifting the zero so the band bottom is at 0.
    def x_SM(self, omega, K=0.0):
        return (omega)


# ----- Parameters (consistent with SM Fig. 2) -----
s = 0.5
J1 = 5         # choose |J| = 1 as energy unit
W = 10.0
t = 1      # W = 4t
#omega_m = 0.01
J0 = 0.05#omega_m / (4.0 * s)   # correct mapping
gamma = 1.0

pol = RealSpaceMagneticPolaron(t, J0, J1, s, gamma)

# Create k and omega grids
k_values = np.linspace(0, np.pi, 200)
omega_values = np.linspace(-6*t, 2*t, 400)
eta = 0.02

# Compute spectral weight on 2D grid
print("Computing spectral weight on 2D grid...")
spectral_2d = np.zeros((len(omega_values), len(k_values)))

for i, k in enumerate(k_values):
    if i % 20 == 0:
        print(f"Progress: {i}/{len(k_values)}")
    for j, omega in enumerate(omega_values):
        spectral_2d[j, i] = pol.spectral_weight(k, omega, eta=eta)

# Create the plot
fig, ax = plt.subplots(figsize=(6, 4.5))

# Plot as 2D image
im = ax.imshow(spectral_2d, 
               extent=[0, 1, omega_values[0]/t, omega_values[-1]/t],
               aspect='auto',
               origin='lower',
               cmap='Reds',  # inverted grayscale like the figure
               vmin=0,
               vmax=np.percentile(spectral_2d, 99))  # clip at 99th percentile for better contrast

# Add colorbar
cbar = plt.colorbar(im, ax=ax)
cbar.set_label(r'$\rho_\downarrow(k,\omega)$')

# Add dashed line at omega=0 (or wherever appropriate)
ax.axhline(-0.75, ls='--', c='grey', lw=2, alpha=0.5)
#plt.plot(k_values/np.pi, q_min_array + E_min_array)
# Labels
ax.set_xlabel(r'$k/\pi$')
ax.set_ylabel(r'$\omega/t$')
#ax.text(0.95, 0.05, '(b)', transform=ax.transAxes, 
#        fontsize=12, verticalalignment='bottom', horizontalalignment='right')
ax.yaxis.set_major_locator(ticker.MultipleLocator(2))
plt.tight_layout()
plt.savefig('spectral_function_2d.png', dpi=200, bbox_inches='tight')
print("Plot saved to outputs/spectral_function_2d.png")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

class RealSpaceMagneticPolaron:
    """
    Real-space magnetic polaron Green's function (one-magnon sector, general s).
    Conventions here:
      - Electron band bottom set to 0: ε_k = 2t(1 - cos K)
      - J1 is taken positive and used as the energy unit |J|
    """
    def __init__(self, t, J_0, J_1, s, gamma=1.0):
        self.t = t
        self.J_0 = J_0
        self.J_1 = J_1
        self.s = s
        self.gamma = gamma

    def epsilon_k(self, K):
        # Band bottom at 0
        return -2.0 * self.t * (np.cos(K))

    def compute_rho(self, z, K):
        """Solve A_- ρ^2 + α ρ + A_+ = 0 and pick |ρ|<1."""
        A_plus  = self.t * np.exp(1j*K) + self.s * self.J_0
        A_minus = self.t * np.exp(-1j*K) + self.s * self.J_0
        alpha = z - 2.0*self.J_0*self.s - 0.5*self.J_1*self.s
        disc = alpha*alpha - 4.0*A_plus*A_minus
        sqrt_disc = np.sqrt(disc + 0j)
        rho1 = (-alpha + sqrt_disc) / (2.0*A_minus)
        rho2 = (-alpha - sqrt_disc) / (2.0*A_minus)
        rho = rho1 if abs(rho1) < abs(rho2) else rho2
        return rho, A_minus

    def compute_G_minus(self, z, K):
        """G(z,K,-) with continued-fraction self-energy."""
        rho, A_minus = self.compute_rho(z, K)
        beta = z - 2.0*self.J_0*self.s - 0.5*self.J_1*(self.s - 1.0)
        vertex_sq = (self.gamma*self.J_1)**2 * self.s / 2.0
        Sigma = vertex_sq / (beta + 2.0*A_minus*rho)
        # With ε_k bottom at 0
        denom = z - self.epsilon_k(K) + 0.5*self.J_1*self.s - Sigma
        return 1.0 / denom

    def spectral_weight(self, K, omega, eta=0.01):
        """ρ(K,ω) = -(1/π) Im G(ω+iη, K, -)."""
        z = omega + 1j*eta
        Gm = self.compute_G_minus(z, K)
        return -np.imag(Gm) / np.pi

    def find_pole(self, K, wmin, wmax, n=6000):
        """Crude scan for an isolated pole (δ-peak)."""
        ws = np.linspace(wmin, wmax, n)
        eta = 1e-4
        reinv = []
        iminv = []
        for w in ws:
            invG = 1.0 / self.compute_G_minus(w+1j*eta, K)
            reinv.append(np.real(invG)); iminv.append(np.imag(invG))
        reinv = np.array(reinv); iminv = np.array(iminv)
        poles = []
        for i in range(len(ws)-1):
            if reinv[i]*reinv[i+1] < 0:
                w0 = ws[i] - reinv[i]*(ws[i+1]-ws[i])/(reinv[i+1]-reinv[i])
                if abs(iminv[i]) < 1e-3:
                    poles.append(w0)
        return min(poles) if poles else None

# -------- Parameters (SM Fig.2-like) --------
s = 1.0/2.0
J = 1.0          # |J|
W = 10.0
omega_m = 0.01
gamma = 0.5

t  = 1                   # W = 4t
J0 = 0.05        # ω_max = 4 J0 s
J1 = 5                        # keep positive; use as unit

print(f"t={t:.3f}, J0={J0:.6f}, J1={J1:.3f}, s={s}, gamma={gamma}")

pol = RealSpaceMagneticPolaron(t, J0, J1, s, gamma)

# -------- Spectrum at K=0 --------
K = 0.0
omega = np.linspace(-6.0, 6.0, 3000)

# Two different broadenings for readability
eta_plot = 0.01   # use larger η to see the continuum shape
rho = np.array([pol.spectral_weight(K, w, eta=eta_plot) for w in omega])

# SM x-axis: x = ω/|J| + s/2
x_SM = omega / J + s/2.0

# Hartree marker for K=0: x_H = s
x_H = s

# Try to locate discrete polaron pole
w_pole = pol.find_pole(K, omega.min(), omega.max(), n=8000)
x_pole = (w_pole/abs(J) + s/2.0) if w_pole is not None else None

# -------- Plot (clipped linear) --------
plt.figure(figsize=(8, 4.5))
plt.plot(x_SM, rho, color='b', lw=2, label=f'η = {eta_plot}')
plt.axvline(x_H, color='r', ls='--', lw=1.3, label='Hartree: x = s')
if x_pole is not None:
    plt.axvline(x_pole, color='k', ls=':', lw=1.2, label='Polaron pole')
plt.xlabel(r'$\omega/|J| + s/2$')
plt.ylabel(r'$\rho_\downarrow(K{=}0,\omega)$')
plt.title(f'$J>0$ (FM), $s={s}$, $W={W}$, $\\omega_m={omega_m}$ (1D TB)')
plt.xlim(-6, 7)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np, math

# Optional sparse backend
try:
    import scipy.sparse as sp
    import scipy.sparse.linalg as spla
    SPARSE_OK = True
except Exception:
    SPARSE_OK = False


class ExcitonMagnonSolver:
    """
    Real-space exciton–magnon Green function on a FM background.

    Unknown vector ordering (by R-blocks):
        [ G(R), G(R,-M), ..., G(R,-1), G(R,0), G(R,1), ..., G(R,M) ] for R=-Rmax..Rmax

    Source (delta) is at the no-magnon entry of R=0 only.
    """

    def __init__(self, t_c, t_v, J0, J1, s, gamma, U, Rmax, M):
        # model params
        self.t_c   = float(t_c)
        self.t_v   = float(t_v)
        self.J0    = float(J0)
        self.J1    = float(J1)
        self.s     = float(s)
        self.gamma = float(gamma)
        self.U     = float(U)

        # truncation
        self.Rmax = int(Rmax)
        self.M    = int(M)

        # sizes
        self.NR   = 2*self.Rmax + 1           # R in [-Rmax, …, Rmax]
        self.Nm   = 2*self.M + 1              # m in [-M, …, M]
        self.block_size = 1 + self.Nm         # [no-magnon] + Nm
        self.Ntot = self.NR * self.block_size

        # handy constants
        self.g_sf   = self.gamma * self.J1 * math.sqrt(self.s/2.0)  # γ J1 √(s/2)
        self.Js_half = 0.5 * self.J1 * self.s                       # J1 s / 2

    # ---------- indexing helpers ----------
    def block_start(self, R):  # start idx of block for this R
        return (R + self.Rmax) * self.block_size

    def idx_no(self, R):       # no-magnon entry in block R
        return self.block_start(R)

    def idx_m(self, R, m):     # one-magnon entry with separation m in block R
        return self.block_start(R) + 1 + (m + self.M)

    def _R_in_range(self, R):
        return (-self.Rmax <= R <= self.Rmax)

    def _m_in_range(self, m):
        return (-self.M <= m <= self.M)

    # ---------- assemble M and e from the exact EOM ----------
    def assemble(self, z, K):
        """
        Build the linear system M(z,K) G = e using the exact real-space EOM.

        Returns
        -------
        M : (Ntot x Ntot) complex matrix (csr if sparse available, dense otherwise)
        e : (Ntot,) complex vector with a single 1.0 at (R=0, no-magnon)
        """
        z = complex(z)
        K = float(K)

        # phases & R-recursion couplings
        exp_iK  = np.exp(1j*K)
        exp_miK = np.exp(-1j*K)
        Bp = self.t_c * exp_iK  + self.t_v * exp_miK  # couples to R-1
        Bm = self.t_c * exp_miK + self.t_v * exp_iK   # couples to R+1

        # on-site diagonals (depend on R only via U δ_{R,0})
        diag_no   = lambda R: z - self.Js_half - (self.U if R == 0 else 0.0)
        diag_m0   = lambda R: z - 2*self.J0*self.s - self.Js_half + 0.5*self.J1 - (self.U if R == 0 else 0.0)
        diag_mne0 = lambda R: z - 2*self.J0*self.s - self.Js_half - (self.U if R == 0 else 0.0)

        # allocate
        if SPARSE_OK:
            rows, cols, data = [], [], []
            def add(i, j, val):
                if val != 0:
                    rows.append(i); cols.append(j); data.append(val)
        else:
            M = np.zeros((self.Ntot, self.Ntot), dtype=np.complex128)
            def add(i, j, val):
                if val != 0:
                    M[i, j] += val

        e = np.zeros(self.Ntot, dtype=np.complex128)

        # fill rows
        for R in range(-self.Rmax, self.Rmax + 1):
            # (i) no-magnon line at this R
            i_no = self.idx_no(R)
            add(i_no, i_no, diag_no(R))
            if self._R_in_range(R - 1): add(i_no, self.idx_no(R - 1),  Bp)       # +B_+ G(R-1)
            if self._R_in_range(R + 1): add(i_no, self.idx_no(R + 1),  Bm)       # +B_- G(R+1)
            add(i_no, self.idx_m(R, 0), -self.g_sf)                               # -γJ1√(s/2) G(R,0)

            # source term δ_{R,0}
            if R == 0:
                e[i_no] = 1.0

            # (ii) m = 0 line at this R
            i_m0 = self.idx_m(R, 0)
            add(i_m0, i_m0, diag_m0(R))
            add(i_m0, i_no, -self.g_sf)                                           # -γJ1√(s/2) G(R)

            # magnon hops along m
            if self._m_in_range(-1): add(i_m0, self.idx_m(R, -1),  self.J0*self.s)
            if self._m_in_range(+1): add(i_m0, self.idx_m(R, +1),  self.J0*self.s)

            # electron hops change both R and m:
            if self._R_in_range(R - 1) and self._m_in_range(-1):
                add(i_m0, self.idx_m(R - 1, -1), self.t_c * exp_miK)
            if self._R_in_range(R + 1) and self._m_in_range(+1):
                add(i_m0, self.idx_m(R + 1, +1), self.t_c * exp_iK)

            # hole hops change only R:
            if self._R_in_range(R - 1):
                add(i_m0, self.idx_m(R - 1, 0), self.t_v * exp_miK)
            if self._R_in_range(R + 1):
                add(i_m0, self.idx_m(R + 1, 0), self.t_v * exp_iK)

            # (iii) m ≠ 0 lines at this R
            for m in range(-self.M, self.M + 1):
                if m == 0:
                    continue
                i_m = self.idx_m(R, m)
                add(i_m, i_m, diag_mne0(R))

                # magnon hops along m:
                if self._m_in_range(m - 1): add(i_m, self.idx_m(R, m - 1), self.J0*self.s)
                if self._m_in_range(m + 1): add(i_m, self.idx_m(R, m + 1), self.J0*self.s)

                # electron hops (shift both R and m):
                if self._R_in_range(R - 1) and self._m_in_range(m - 1):
                    add(i_m, self.idx_m(R - 1, m - 1), self.t_c * exp_miK)
                if self._R_in_range(R + 1) and self._m_in_range(m + 1):
                    add(i_m, self.idx_m(R + 1, m + 1), self.t_c * exp_iK)

                # hole hops (shift R only):
                if self._R_in_range(R - 1):
                    add(i_m, self.idx_m(R - 1, m), self.t_v * exp_miK)
                if self._R_in_range(R + 1):
                    add(i_m, self.idx_m(R + 1, m), self.t_v * exp_iK)

        if SPARSE_OK:
            M = sp.csr_matrix((data, (rows, cols)), shape=(self.Ntot, self.Ntot), dtype=np.complex128)

        return M, e

    # ---------- linear solve ----------
    def solve(self, z, K):
        M, e = self.assemble(z, K)
        if SPARSE_OK:
            return spla.spsolve(M, e)
        return np.linalg.solve(M, e)

    # ---------- accessors ----------
    def get_G(self, Gvec, R, m=None):
        """Return G(z,K;R) if m is None; else G(z,K;R,m)."""
        return Gvec[self.idx_no(R) if m is None else self.idx_m(R, m)]

    # ---------- convenience: spectral weight at (R=0, no-magnon) ----------
    def spectral_weight(self, K, omega, eta):
        z = omega + 1j*eta
        Gvec = self.solve(z, K)
        return -np.imag(self.get_G(Gvec, R=0)) / np.pi
```

```python
import numpy as np, math

# Optional sparse backend
try:
    import scipy.sparse as sp
    import scipy.sparse.linalg as spla
    SPARSE_OK = True
except Exception:
    SPARSE_OK = False


class ExcitonMagnonSolver:
    """
    Real-space exciton–magnon Green function on a FM background with a band gap.

    Bands:
        ε_c(k) = 2 t_c (1 - cos k)      (electron)
        ε_v(k) = 2 t_v cos k + Delta    (valence electron)
        ε_h(k) = -ε_v(k)                (hole)

    Unknown vector ordering (by R-blocks):
        [ G(R), G(R,-M), ..., G(R,-1), G(R,0), G(R,1), ..., G(R,M) ] for R=-Rmax..Rmax
    """

    def __init__(self, t_c, t_v, Delta, J0, J1, s, gamma, U, Rmax, M):
        # model params
        self.t_c   = float(t_c)
        self.t_v   = float(t_v)
        self.Delta = float(Delta)        # < 0 to push valence down
        self.J0    = float(J0)
        self.J1    = float(J1)
        self.s     = float(s)
        self.gamma = float(gamma)
        self.U     = float(U)

        # truncation
        self.Rmax = int(Rmax)
        self.M    = int(M)

        # sizes
        self.NR   = 2*self.Rmax + 1
        self.Nm   = 2*self.M + 1
        self.block_size = 1 + self.Nm
        self.Ntot = self.NR * self.block_size

        # handy constants
        self.g_sf    = self.gamma * self.J1 * math.sqrt(self.s/2.0)  # γ J1 √(s/2)
        self.Js_half = 0.5 * self.J1 * self.s                        # J1 s / 2

        # constant pair offset from band definitions:
        # ε_c(k) has +2 t_c; ε_h(k) adds -Δ  ⇒ total constant E_off = 2 t_c - Δ
        self.E_off = 2.0*self.t_c - self.Delta

    # ---------- reference dispersions / gap ----------
    def eps_c(self, k): return 2.0*self.t_c*(1.0 - np.cos(k))
    def eps_v(self, k): return 2.0*self.t_v*np.cos(k) + self.Delta
    def eps_h(self, k): return -self.eps_v(k)
    def direct_gap(self): return -(2.0*self.t_v + self.Delta)  # at K=0

    # ---------- indexing helpers ----------
    def block_start(self, R):  return (R + self.Rmax) * self.block_size
    def idx_no(self, R):       return self.block_start(R)
    def idx_m(self, R, m):     return self.block_start(R) + 1 + (m + self.M)
    def _R_in_range(self, R):  return (-self.Rmax <= R <= self.Rmax)
    def _m_in_range(self, m):  return (-self.M    <= m <= self.M)

    # ---------- assemble M and e from the exact EOM ----------
    def assemble(self, z, K):
        """
        Build the linear system M(z,K) G = e using the exact real-space EOM.

        Returns
        -------
        M : (Ntot x Ntot) complex matrix (csr if sparse available, dense otherwise)
        e : (Ntot,) complex vector with a single 1.0 at (R=0, no-magnon)
        """
        z = complex(z); K = float(K)

        exp_iK  = np.exp(1j*K)
        exp_miK = np.exp(-1j*K)

        # NO-MAGNON couplings along R:
        #  (H_c + H_v)|R> = -[ t_c e^{+iK} - t_v e^{-iK} ] |R-1>
        #                 -[ t_c e^{-iK} - t_v e^{+iK} ] |R+1>
        Bp = self.t_c*exp_iK  - self.t_v*exp_miK   # to R-1
        Bm = self.t_c*exp_miK - self.t_v*exp_iK    # to R+1

        # diagonals (all pick up -E_off in (z-H))
        diag_no   = lambda R: z - self.E_off - self.Js_half - (self.U if R == 0 else 0.0)
        diag_m0   = lambda R: z - self.E_off - 2*self.J0*self.s - self.Js_half + 0.5*self.J1 \
                              - (self.U if R == 0 else 0.0)
        diag_mne0 = lambda R: z - self.E_off - 2*self.J0*self.s - self.Js_half \
                              - (self.U if R == 0 else 0.0)

        # storage
        if SPARSE_OK:
            rows, cols, data = [], [], []
            def add(i, j, val):
                if val != 0: rows.append(i); cols.append(j); data.append(val)
        else:
            M = np.zeros((self.Ntot, self.Ntot), dtype=np.complex128)
            def add(i, j, val):
                if val != 0: M[i, j] += val

        e = np.zeros(self.Ntot, dtype=np.complex128)

        # fill rows
        for R in range(-self.Rmax, self.Rmax + 1):
            # (i) no-magnon line
            i_no = self.idx_no(R)
            add(i_no, i_no, diag_no(R))
            if self._R_in_range(R - 1): add(i_no, self.idx_no(R - 1),  Bp)
            if self._R_in_range(R + 1): add(i_no, self.idx_no(R + 1),  Bm)
            add(i_no, self.idx_m(R, 0), -self.g_sf)  # -γJ1√(s/2) G(R,0)
            if R == 0: e[i_no] = 1.0                # source

            # (ii) m = 0 line
            i_m0 = self.idx_m(R, 0)
            add(i_m0, i_m0, diag_m0(R))
            add(i_m0, i_no, -self.g_sf)

            # magnon hops along m
            if self._m_in_range(-1): add(i_m0, self.idx_m(R, -1),  self.J0*self.s)
            if self._m_in_range(+1): add(i_m0, self.idx_m(R, +1),  self.J0*self.s)

            # electron hops (from -t_c ... => signs as before)
            if self._R_in_range(R - 1) and self._m_in_range(-1): add(i_m0, self.idx_m(R - 1, -1), self.t_c*exp_miK)
            if self._R_in_range(R + 1) and self._m_in_range(+1): add(i_m0, self.idx_m(R + 1, +1), self.t_c*exp_iK)

            # hole hops (NOW opposite sign compared with old code, because H_v = +t_v ...)
            if self._R_in_range(R - 1): add(i_m0, self.idx_m(R - 1,  0), -self.t_v*exp_miK)
            if self._R_in_range(R + 1): add(i_m0, self.idx_m(R + 1,  0), -self.t_v*exp_iK)

            # (iii) m ≠ 0 lines
            for m in range(-self.M, self.M + 1):
                if m == 0: continue
                i_m = self.idx_m(R, m)
                add(i_m, i_m, diag_mne0(R))

                # magnon hops along m
                if self._m_in_range(m - 1): add(i_m, self.idx_m(R, m - 1), self.J0*self.s)
                if self._m_in_range(m + 1): add(i_m, self.idx_m(R, m + 1), self.J0*self.s)

                # electron hops (−t_c ...)
                if self._R_in_range(R - 1) and self._m_in_range(m - 1): add(i_m, self.idx_m(R - 1, m - 1), self.t_c*exp_miK)
                if self._R_in_range(R + 1) and self._m_in_range(m + 1): add(i_m, self.idx_m(R + 1, m + 1), self.t_c*exp_iK)

                # hole hops (sign flipped)
                if self._R_in_range(R - 1): add(i_m, self.idx_m(R - 1, m), -self.t_v*exp_miK)
                if self._R_in_range(R + 1): add(i_m, self.idx_m(R + 1, m), -self.t_v*exp_iK)

        if SPARSE_OK:
            M = sp.csr_matrix((data, (rows, cols)), shape=(self.Ntot, self.Ntot), dtype=np.complex128)

        return M, e

    # ---------- accessors ----------
    def get_G(self, Gvec, R, m=None):
        return Gvec[self.idx_no(R) if m is None else self.idx_m(R, m)]

    # ---------- flexible solver / spectra ----------
    def _zeros(self): 
        return np.zeros(self.Ntot, dtype=np.complex128)

    def solve(self, z, K, source='no', Rsrc=0, msrc=0):
        M, _ = self.assemble(z, K)
        e = self._zeros()
        if source == 'no':
            e[self.idx_no(Rsrc)] = 1.0
        elif source == 'm':
            e[self.idx_m(Rsrc, msrc)] = 1.0
        else:
            raise ValueError('source must be "no" or "m"')
        if SPARSE_OK: 
            return spla.spsolve(M, e)
        return np.linalg.solve(M, e)

    def spectral(self, K, omega, eta, R=0, m=None, *, source='no', Rsrc=0, msrc=0):
        z = omega + 1j*eta
        Gvec = self.solve(z, K, source=source, Rsrc=Rsrc, msrc=msrc)
        Gij  = self.get_G(Gvec, R=R, m=m)
        return -np.imag(Gij) / np.pi
```

```python
solver = ExcitonMagnonSolver(
    t_c=1.0, t_v= -1.0, Delta = 0, J0=0.05, J1=20, s=0.5, gamma=1.0, U=1,
    Rmax=20, M=10
)
omegas = np.linspace(-20,15,100)
A_no = np.array([solver.spectral(0.0, w, 0.1, R=0, m=None,
                                 source='no', Rsrc=0)
                 for w in omegas])
```
